# COVID-19 EDA — CodeAlpha Task 2

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('owid-covid-data.csv')
print(df.shape)
```

```python
df.describe()
```

```python
# Checking the missing values
df.isnull().sum()
```

```python
#conversion from text to datetime
df['date'] = pd.to_datetime(df['date'])

## removing world and continent rows, keeping only countries
df = df[df['continent'].notna()]

print('Cleaned! Countries:', df['location'].nunique())
```

```python
# --- CHART 1: Global Total Cases Over Time ---
# Total cases on a particular date
global_cases = df.groupby('date')['total_cases'].sum()
plt.plot(global_cases.index, global_cases.values)
plt.title('Global Total COVID-19 Cases Over Time')
plt.xlabel('Date')
plt.ylabel('Total Cases')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# CHART 2:Global Total Deaths Over Time

global_deaths = df.groupby('date')['total_deaths'].sum()

plt.plot(global_deaths.index, global_deaths.values, color='red')
plt.title('Global Total COVID-19 Deaths Over Time')
plt.xlabel('Date')
plt.ylabel('Total Deaths')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# CHART 3:Top 10 Countries by Total Cases

# Get the latest record per country
latest = df.sort_values('date').groupby('location').last().reset_index()

# Pick top 10 by total cases
top10 = latest.nlargest(10, 'total_cases')

plt.bar(top10['location'], top10['total_cases'])
plt.title('Top 10 Countries by Total Cases')
plt.xlabel('Country')
plt.ylabel('Total Cases')
plt.xticks(rotation=45)
plt.show()
```

```python
#CHART 4Top 10 Countries by Total Deaths

top10_deaths = latest.nlargest(10, 'total_deaths')

plt.bar(top10_deaths['location'], top10_deaths['total_deaths'], color='red')
plt.title('Top 10 Countries by Total Deaths')
plt.xlabel('Country')
plt.ylabel('Total Deaths')
plt.xticks(rotation=45)
plt.show()
```

```python
#CHART 5:otal Cases by Continent
continent_cases = latest.groupby('continent')['total_cases'].sum()
plt.pie(continent_cases.values, labels=continent_cases.index, autopct='%1.1f%%')
plt.title('COVID-19 Cases Share by Continent')
plt.show()
```

```python
# CHART 6:Daily New Cases
daily_cases = df.groupby('date')['new_cases'].sum()
# smoothing the data using 7 day average
rolling = daily_cases.rolling(7).mean()
plt.plot(daily_cases.index, daily_cases.values, alpha=0.3, label='Daily Cases')
plt.plot(rolling.index, rolling.values, color='blue', label='7-Day Average')
plt.title('Global Daily New Cases')
plt.xlabel('Date')
plt.ylabel('New Cases')
plt.xticks(rotation=45)
plt.tight_layout()
plt.legend()
plt.show()
```

```python
#CHART 7:Top 10 Countries by Vaccinations
if 'people_fully_vaccinated' in latest.columns:
    vacc = latest.dropna(subset=['people_fully_vaccinated'])
    top10_vacc = vacc.nlargest(10, 'people_fully_vaccinated')
    plt.bar(top10_vacc['location'], top10_vacc['people_fully_vaccinated'], color='green')
    plt.title('Top 10 Countries by People Fully Vaccinated')
    plt.xlabel('Country')
    plt.ylabel('People Fully Vaccinated')
    plt.xticks(rotation=45)
    plt.show()
else:
    print('Vaccination data not available in this dataset version')
```

```python
#CHART 8: Cases vs Deaths Scatter Plot
scatter = latest.dropna(subset=['total_cases', 'total_deaths'])
plt.scatter(scatter['total_cases'], scatter['total_deaths'], alpha=0.5)
plt.title('Total Cases vs Total Deaths by Country')
plt.xlabel('Total Cases')
plt.ylabel('Total Deaths')
plt.show()
```

```python
# Conclusion
print('Total Global Cases  :', latest['total_cases'].sum())
print('Total Global Deaths :', latest['total_deaths'].sum())
print('Country with most cases  :', latest.loc[latest['total_cases'].idxmax(), 'location'])
print('Country with most deaths :', latest.loc[latest['total_deaths'].idxmax(), 'location'])
```
